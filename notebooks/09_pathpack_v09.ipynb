# PathPack v0.9

Три прогона на тех же 20 dev-задачах. v0.4 не заменён: ни одна конфигурация не обошла его сразу по File, Function, Complete и Changed Line Recall. Числа читаются из `output/v09_dev`, `output/v09_name_dev` и `output/v09_splice_dev`. Конфиги после прогонов не менялись. Holdout не оценивался.


```python
from pathlib import Path
import pandas as pd
ROOT = Path.cwd().resolve()
if not (ROOT / 'output').exists():
    ROOT = ROOT.parent
runs = {
    'pool+coverage': ROOT / 'output/v09_dev/summary.csv',
    'name prior': ROOT / 'output/v09_name_dev/summary.csv',
    'name splice': ROOT / 'output/v09_splice_dev/summary.csv',
}
cols = ['method', 'file_recall', 'function_recall', 'complete_function_recall', 'changed_line_recall', 'latency_sec']
for name, path in runs.items():
    frame = pd.read_csv(path)
    view = frame.loc[frame['budget'] == 2048, cols].copy()
    print('\n', name)
    print(view.round(3).to_string(index=False))
```

```python
metrics = ['file_recall', 'function_recall', 'complete_function_recall', 'changed_line_recall']
pairs = {
    'pool+coverage': ROOT / 'output/v09_dev/comparisons.csv',
    'name prior': ROOT / 'output/v09_name_dev/comparisons.csv',
    'name splice': ROOT / 'output/v09_splice_dev/comparisons.csv',
}
for name, path in pairs.items():
    frame = pd.read_csv(path)
    view = frame[(frame.candidate == 'pathpack-v9') & (frame.baseline == 'pathpack-v4') & (frame.budget == 2048) & frame.metric.isin(metrics)]
    print('\n', name, 'v9 − v4 @ 2048')
    print(view[['metric', 'n', 'difference', 'ci95', 'wins', 'ties', 'losses']].to_string(index=False))
```
